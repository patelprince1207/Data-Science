```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
cities = ["Bangalore", "Chennai", "Delhi", "Hyderabad", "Kolkata", "Mumbai"]
```

```python
df_list = []
for city in cities:
    temp = pd.read_csv(f"{city}.csv")
    temp ["city"] = city
    df_list.append(temp)
```

```python
df = pd.concat(df_list, ignore_index=True)
print(df.shape)
df.head()
```
