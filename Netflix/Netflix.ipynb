```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("netflix_titles.csv")

# -------------------------------
# Basic Information
# -------------------------------

print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())

# -------------------------------
# Movies vs TV Shows
# -------------------------------

type_count = df['type'].value_counts()

print("\nMovies vs TV Shows:")
print(type_count)

plt.figure(figsize=(6,4))
sns.countplot(x='type', data=df)
plt.title("Movies vs TV Shows")
plt.show()

# -------------------------------
# Top 10 Countries
# -------------------------------

country_data = (
    df['country']
    .dropna()
    .str.split(', ')
    .explode()
)

top_countries = country_data.value_counts().head(10)

print("\nTop 10 Countries:")
print(top_countries)

plt.figure(figsize=(10,5))
top_countries.plot(kind='bar')
plt.title("Top 10 Countries by Content")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()

# -------------------------------
# Rating Distribution
# -------------------------------

rating_count = df['rating'].value_counts()

print("\nRating Distribution:")
print(rating_count)

plt.figure(figsize=(10,5))
sns.countplot(
    y='rating',
    data=df,
    order=df['rating'].value_counts().index
)
plt.title("Content Ratings")
plt.show()

# -------------------------------
# Top 10 Genres
# -------------------------------

genre_data = (
    df['listed_in']
    .str.split(', ')
    .explode()
)

top_genres = genre_data.value_counts().head(10)

print("\nTop Genres:")
print(top_genres)

plt.figure(figsize=(10,5))
top_genres.plot(kind='bar')
plt.title("Top 10 Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.show()

# -------------------------------
# Year-wise Content Growth
# -------------------------------

year_growth = (
    df['release_year']
    .value_counts()
    .sort_index()
)

print("\nYear-wise Content Growth:")
print(year_growth.tail())

plt.figure(figsize=(12,5))
year_growth.plot(kind='line')
plt.title("Netflix Content Growth Over Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()

# -------------------------------
# Top 10 Directors
# -------------------------------

top_directors = (
    df['director']
    .dropna()
    .value_counts()
    .head(10)
)

print("\nTop Directors:")
print(top_directors)

plt.figure(figsize=(10,5))
top_directors.plot(kind='bar')
plt.title("Top 10 Directors")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.show()

print("\nNetflix Data Analysis Completed Successfully!")
```
